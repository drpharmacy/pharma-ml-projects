**Breast Cancer Classification Using Decision Tree Machine Learning Model** or **Prediction of Breast Cancer Diagnosis Using a Decision Tree Classifier** or **Machine Learning-Based Breast Cancer Classification Using the Decision Tree Algorithm**

```python
# Import Libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
# Dataset URL
url = "https://raw.githubusercontent.com/drpharmacy/pharma-ml-projects/refs/heads/main/Excel_Files/1%20breastcancer%20.csv"

# Load Dataset
df = pd.read_csv(url)

# Display first five rows
df.head()
```

**Meaning of Code Block 1**

This code block imports the required Python libraries, loads the Breast Cancer dataset directly from the GitHub repository using its raw URL into a Pandas DataFrame, and displays the first five records. This helps verify that the dataset has been imported successfully and allows us to understand its structure before developing the Decision Tree machine learning model.

```python
# Select Input Features (X)
X = df.drop(columns=["id", "diagnosis", "Unnamed: 32"])

# Select Target Variable (Y)
y = df["diagnosis"]
```

**Meaning of Code Block 2**

This code block separates the dataset into input features (X) and the target variable (Y). The input features (X) consist of 30 numerical measurements of breast cell nuclei that are used to predict the diagnosis. The target variable (Y) is the diagnosis column, where M represents a malignant (cancerous) tumor and B represents a benign (non-cancerous) tumor. The Decision Tree model learns the relationship between the input features and the diagnosis to classify new patients accurately.

```python
# Split Dataset into Training and Testing Sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

**Meaning of Code Block 3**

This code block divides the dataset into two parts: a training dataset and a testing dataset. The training dataset (80%) is used to train the Decision Tree model, while the testing dataset (20%) is used to evaluate its performance on unseen patient data. This ensures that the model is tested on data it has not previously seen, providing a reliable assessment of its predictive accuracy. The random_state=42 parameter ensures that the dataset is split in the same way every time the program is executed, making the results reproducible.

```python
# Create the Decision Tree Model
model = DecisionTreeClassifier(random_state=42)
```

**Meaning of Code Block 4**

This code block creates a Decision Tree machine learning model, which is a supervised learning algorithm used for classification. In this practical, the model is designed to classify breast tumors into two categories: Malignant (M) and Benign (B). The random_state=42 parameter ensures that the Decision Tree is generated in the same way each time the program is executed, making the results consistent and reproducible.

```python
# Train the Decision Tree Model
model.fit(X_train, y_train)
```

**Meaning of Code Block 5**

This code block trains the Decision Tree machine learning model using the training dataset. During training, the model learns the relationship between the input features (breast cell measurements) and the diagnosis (malignant or benign). The trained model identifies patterns within the data, enabling it to classify new patients accurately based on their breast cell characteristics.

```python
# Predict the Diagnosis
y_pred = model.predict(X_test)
```

**Meaning of Code Block 6**

This code block uses the trained Decision Tree model to predict the diagnosis of breast tumors for the patients in the testing dataset. The predicted diagnoses are stored in the variable y_pred, where each prediction is classified as either Malignant (M) or Benign (B). These predicted values are later compared with the actual diagnoses to evaluate the accuracy and performance of the Decision Tree machine learning model.

```python
# Compare Actual and Predicted Diagnosis
results = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})

results.head(10)
```

**Meaning of Code Block 7**

This code block creates a table that displays the actual diagnosis and the predicted diagnosis for the first ten patients in the testing dataset. Comparing these values helps determine whether the Decision Tree model has correctly classified each patient as Malignant (M) or Benign (B). This provides a simple visual assessment of the model's prediction performance before calculating detailed evaluation metrics such as accuracy, confusion matrix, and classification report.

```python
# Evaluate the Decision Tree Model
print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))
```

**Meaning of Code Block 8** **Evaluate Model (Accuracy, Confusion Matrix, Classification Report)**

This code block evaluates the performance of the Decision Tree machine learning model by calculating the accuracy, generating a confusion matrix, and displaying a classification report. These evaluation metrics indicate how well the model classifies breast tumors as Malignant (M) or Benign (B). The accuracy represents the overall percentage of correct predictions, the confusion matrix summarizes the correct and incorrect classifications, and the classification report provides important performance measures such as precision, recall, and F1-score. Together, these metrics help determine the reliability and effectiveness of the Decision Tree model in breast cancer classification.

```python
# Import Libraries for Visualization
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

# Plot the Decision Tree
plt.figure(figsize=(20,10))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=["Benign", "Malignant"],
    filled=True,
    rounded=True,
    fontsize=8
)

plt.title("Decision Tree for Breast Cancer Classification")
plt.show()
```

**Meaning of Code Block 9**

This code block creates a graphical representation of the Decision Tree model. The diagram illustrates how the model makes classification decisions by splitting the data based on different breast cell features. Each node represents a decision based on a specific feature, while the terminal nodes (leaf nodes) represent the final prediction of either Benign or Malignant. The colored nodes indicate the predicted class, making it easier to understand how the Decision Tree reaches its classification decisions.

This is the standard visualization used for teaching Decision Trees.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Create Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

# Plot Confusion Matrix
plt.figure(figsize=(6,5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Benign", "Malignant"],
    yticklabels=["Benign", "Malignant"]
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("Confusion Matrix for Breast Cancer Classification")
plt.show()
```

**Meaning of Code Block 10**

This code block creates a Confusion Matrix Heatmap, which provides a graphical representation of the **model's classification performance**. The heatmap compares the actual diagnosis with the predicted diagnosis and displays the number of correct and incorrect predictions for each class. A higher number of values along the diagonal indicates better model performance, while values outside the diagonal represent misclassified cases. This visualization helps students quickly understand the accuracy and reliability of the Decision Tree model.

This is the most commonly used evaluation graph in classification problems. After this, you can optionally add ROC Curve and Feature Importance as advanced visualizations.

```python
# Import Libraries
import matplotlib.pyplot as plt
import pandas as pd

# Create a DataFrame of Feature Importances
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

# Sort Features by Importance
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

# Plot Feature Importance
plt.figure(figsize=(10,8))
plt.barh(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xlabel("Feature Importance")
plt.ylabel("Features")
plt.title("Feature Importance in Decision Tree Model")
plt.gca().invert_yaxis()

plt.show()
```

**Meaning of Code Block 11**

This code block calculates and displays the **importance of each input feature **used by the Decision Tree model during classification. The features are ranked according to their contribution to the prediction of breast cancer diagnosis. The horizontal bar graph helps identify which characteristics of breast cells have the greatest influence on the model's decision-making process. Features with longer bars have a greater impact on classification, while shorter bars contribute less to the prediction. This visualization improves the interpretability of the machine learning model by highlighting the most influential predictors.

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score

# Convert Diagnosis to Numeric Values
y_test_binary = y_test.map({"B": 0, "M": 1})

# Get Prediction Probabilities
y_prob = model.predict_proba(X_test)[:, 1]

# Calculate ROC Curve
fpr, tpr, thresholds = roc_curve(y_test_binary, y_prob)

# Calculate Area Under the Curve (AUC)
auc = roc_auc_score(y_test_binary, y_prob)

# Plot ROC Curve
plt.figure(figsize=(6,6))

plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")
plt.plot([0,1], [0,1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve for Breast Cancer Classification")
plt.legend(loc="lower right")

plt.show()
```

**Meaning of Code Block 12**

This code block creates the Receiver Operating Characteristic (ROC) Curve, which evaluates the classification performance of the Decision Tree model at different decision thresholds. The ROC curve plots the True Positive Rate (Sensitivity) against the False Positive Rate (1 − Specificity). It also calculates the Area Under the Curve (AUC), which summarizes the model's ability to distinguish between malignant and benign breast tumors. An AUC value closer to 1.0 indicates excellent classification performance, while a value closer to 0.5 suggests performance similar to random guessing.

This is the standard ROC curve used in machine learning research and biomedical classification studies.

**Meaning of Code Block 13**

This code block creates a Class Distribution Bar Chart that **displays the number of patients in each diagnosis category of the breast cancer dataset**. The graph shows how many samples belong to the Benign (B) class and how many belong to the Malignant (M) class. This visualization helps determine whether the dataset is balanced or imbalanced before training the Decision Tree model. Understanding the class distribution is important because a highly imbalanced dataset can affect the performance and reliability of a machine learning model.

```python
# Import Libraries
import matplotlib.pyplot as plt
import pandas as pd

# Count Actual and Predicted Classes
actual_counts = y_test.value_counts().sort_index()
predicted_counts = pd.Series(y_pred).value_counts().sort_index()

# Plot Bar Chart
plt.figure(figsize=(7,5))

bar_width = 0.35
x = range(len(actual_counts))

plt.bar(x, actual_counts.values, width=bar_width, label="Actual")
plt.bar([i + bar_width for i in x], predicted_counts.values, width=bar_width, label="Predicted")

plt.xticks(
    [i + bar_width/2 for i in x],
    actual_counts.index
)

plt.xlabel("Diagnosis")
plt.ylabel("Number of Patients")
plt.title("Actual vs Predicted Breast Cancer Diagnosis")
plt.legend()

plt.show()
```

**meaning of block 14**

This code block creates a Predicted vs Actual Bar Chart, which compares the number of actual and predicted breast cancer diagnoses in the testing dataset. The graph displays two bars for each diagnosis category (Benign and Malignant): one representing the true number of patients and the other representing the number predicted by the Decision Tree model. If the heights of the bars are similar, it indicates that the model has made accurate predictions. This visualization provides a simple and intuitive way to assess the model's classification performance.
