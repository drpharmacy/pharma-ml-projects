```python
import matplotlib.pyplot as plt
import numpy as np
```

**Line** **Plot**
Used to show trends over time or continuous data.

```python
Days= ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
Sugar = [110,120,118,135,128,130,111]

plt.plot(Days,Sugar)
plt.title("Blood Sugar trend over week")
plt.xlabel("Days")
plt.ylabel("Sugar")
plt.show()
```

**Scatter Plot**
Used to show the relationship between two variables.

```python
height = [150, 155, 160, 165, 170, 175, 180]
weight = [45, 50, 55, 60, 68, 72, 80]
plt.scatter(height, weight)
plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight(kg)")
plt.grid(True)
plt.show()
```

**3. Bar Chart**

Used to compare different categories.

```python
categories = ["A", "B", "C", "D"]
values = [10, 20, 15, 25]

plt.bar(categories, values)
plt.show()
```

**4. Histogram**
Used to display the distribution of data.

```python
weights= [40,50,52,55,56,58,60,62,65,92]
plt.boxplot(weights)
plt.show()
```

```python
data = [1, 2, 2, 3, 3, 3, 4, 5, 5]

plt.hist(data)
plt.show()
```

**5. Pie Chart**

Shows the proportion of each category.

```python
sizes = [30, 25, 20, 25]
labels = ["A", "B", "C", "D"]

plt.pie(sizes, labels=labels, autopct="%1.1f%%")
plt.show()
```

**Adding a Legend**

```python
x = [1, 2, 3, 4]
y1 = [2, 4, 6, 8]
y2 = [1, 3, 5, 7]

plt.plot(x, y1, label="Product A")
plt.plot(x, y2, label="Product B")

plt.legend()
plt.show()
```

**Customizing Plots**

You can customize plots using various parameters:

```python
plt.plot(
    x,
    y1,
    marker="o",
    linestyle="--",
    linewidth=2
)

plt.show()
```

**Saving a Figure**

```python
plt.plot(x, y1)
plt.savefig("plot.png")
plt.show()
```

Create plots and visualize pharmaceutical datasets -
concentration-time curves for oral and IV administration,
ADR reporting rates across drugs, dissolution profiles.

```python
import matplotlib.pyplot as plt

# Time (hours)
time = [0, 1, 2, 3, 4, 5, 6, 8, 10]

# Drug concentration (mg/L)
oral = [0, 5, 10, 12, 10, 8, 6, 3, 1]
iv = [20, 18, 15, 12, 9, 7, 5, 2, 1]

plt.plot(time, oral, marker='o', label='Oral Administration')
plt.plot(time, iv, marker='s', label='IV Administration')

plt.title("Drug Concentration-Time Curve")
plt.xlabel("Time (Hours)")
plt.ylabel("Drug Concentration (mg/L)")
plt.legend()
plt.grid(True)

plt.show()
```

2. ADR (Adverse Drug Reaction) Reporting Rates Across Drugs

```python
import matplotlib.pyplot as plt

drugs = ["Paracetamol", "Ibuprofen", "Amoxicillin", "Metformin", "Aspirin"]
adr = [12, 18, 10, 7, 15]

plt.bar(drugs, adr)

plt.title("ADR Reporting Rates Across Drugs")
plt.xlabel("Drug Name")
plt.ylabel("ADR Reports")

plt.grid(axis='y')
plt.show()
```

3. Dissolution Profile

```python
import matplotlib.pyplot as plt

# Time (minutes)
time = [0, 10, 20, 30, 40, 50, 60]

# Percentage drug dissolved
drugA = [0, 25, 45, 65, 80, 92, 100]
drugB = [0, 20, 38, 55, 70, 82, 95]

plt.plot(time, drugA, marker='o', label='Drug A')
plt.plot(time, drugB, marker='s', label='Drug B')

plt.title("Dissolution Profile")
plt.xlabel("Time (Minutes)")
plt.ylabel("Drug Dissolved (%)")

plt.legend()
plt.grid(True)

plt.show()
```

```python
from scipy.integrate import trapezoid

# Time (hours)
time = [0, 1, 2, 3, 4, 5, 6, 8, 10]

# Drug concentration (mg/L)
oral = [0, 5, 10, 12, 10, 8, 6, 3, 1]
iv = [20, 18, 15, 12, 9, 7, 5, 2, 1]

# Calculate AUC for Oral Administration
auc_oral = trapezoid(oral, time)
print(f"AUC for Oral Administration: {auc_oral:.2f} mg*hr/L")

# Calculate AUC for IV Administration
auc_iv = trapezoid(iv, time)
print(f"AUC for IV Administration: {auc_iv:.2f} mg*hr/L")
```
