```python
# Import Pandas
import pandas as pd

# GitHub RAW URL of the Excel file
url = "https://raw.githubusercontent.com/drpharmacy/pharma-ml-projects/main/Excel_Files/Pharmacognosy_ML_Master_Dataset.xlsx"

# Load Excel file into a Pandas DataFrame
df = pd.read_excel(url)

# Display the first 5 rows
df.head()
```

**Meaning of block 1**: This code imports the Pandas library and gives it the short name pd, stores the RAW GitHub location of our master Excel file in the variable url, and uses pd.read_excel() to read the Excel file directly from GitHub into a Pandas DataFrame called df. The command df.head() displays the first five rows so that we can confirm that the dataset has been loaded correctly. This same df will be used for our subsequent data-handling and machine-learning operations.

```python
# Block 2: Inspecting the DataFrame Structure

print("Shape of dataset:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataFrame information:")
df.info()
```

**Meaning of Code Block 2**

This code block examines the basic structure of the dataset. df.shape displays the number of rows and columns, df.columns.tolist() displays all column names, df.dtypes shows the data type of each column, and df.info() provides a concise summary of the DataFrame, including the number of records, columns, non-null values, and data types. These commands help us understand the dataset before performing data cleaning, preprocessing, visualization, and machine learning.

```python
# Block 3: Viewing the Dataset

print("First 5 rows:")
display(df.head())

print("\nFirst 10 rows:")
display(df.head(10))

print("\nLast 5 rows:")
display(df.tail())

print("\nLast 10 rows:")
display(df.tail(10))
```

**Meaning of Code Block 3**

This code block is used to view records from the beginning and end of the DataFrame. df.head() displays the first five rows by default, while df.head(10) displays the first ten rows. Similarly, df.tail() displays the last five rows and df.tail(10) displays the last ten rows. The display() function provides a clean tabular representation in Google Colab. These commands are useful for quickly checking whether the dataset has loaded correctly and for inspecting the actual records before further analysis.

```python
# Block 4: Selecting Columns from the DataFrame

# Select a single column
print("Plant names:")
display(df["Plant_Name"])

# Select multiple columns
print("Selected columns:")
display(df[["Plant_Name", "Family", "Plant_Part", "Extraction_Method"]])

# Check the data type of a selected column
print("Data type of Extraction Yield:")
print(df["Extraction_Yield_pct"].dtype)
```

**Meaning of Code Block 4**

This code block teaches how to select individual or multiple columns from a Pandas DataFrame. df["Plant_Name"] selects one column, while df[["Plant_Name", "Family", "Plant_Part", "Extraction_Method"]] selects several columns together and returns them as a new DataFrame. The dtype command shows the data type of a particular column, which is important for deciding how that variable can be used during data analysis and machine learning.

```python
# Block 5: Selecting Rows and Specific Data

# Select the first row
print("First row:")
display(df.iloc[0])

# Select the first five rows
print("First five rows:")
display(df.iloc[0:5])

# Select rows 0 to 4 and columns 0 to 3
print("Selected rows and columns:")
display(df.iloc[0:5, 0:4])

# Select a specific cell
print("Value of first row, Plant Name column:")
print(df.iloc[0]["Plant_Name"])
```

**Meaning of Code Block 5**

This code block demonstrates how to select specific rows and individual data values using Pandas. df.iloc[0] selects the first row, df.iloc[0:5] selects the first five rows, and df.iloc[0:5, 0:4] selects rows 0 to 4 and columns 0 to 3 using integer-based indexing. The expression df.iloc[0]["Plant_Name"] retrieves a particular value from the DataFrame. Understanding row and column selection is essential for examining, manipulating, and preparing datasets for machine learning.

```python
# Block 6: Filtering Rows Using Conditions

# Select samples with extraction yield greater than 25%
high_yield = df[df["Extraction_Yield_pct"] > 25]
display(high_yield)

# Select samples extracted using ethanol
ethanol_samples = df[df["Solvent"] == "Ethanol"]
display(ethanol_samples)

# Select samples with antioxidant activity greater than or equal to 60%
active_samples = df[df["Antioxidant_Activity_pct"] >= 60]
display(active_samples)

# Apply two conditions together
selected_samples = df[
    (df["Extraction_Yield_pct"] > 25) &
    (df["Antioxidant_Activity_pct"] >= 60)
]
display(selected_samples)
```

**Meaning of Code Block 6**

This code block demonstrates how to filter rows from a DataFrame according to specific conditions. Pandas allows us to select only the records that satisfy a particular criterion, such as extraction yield greater than 25%, samples extracted using ethanol, or antioxidant activity of at least 60%. Multiple conditions can also be combined using operators such as & for AND and | for OR. Filtering is an important data-handling operation because it allows researchers to focus on specific samples or conditions during exploratory analysis and subsequent machine-learning workflows.

```python
# Block 7: Selecting Rows and Columns Using loc

# Select specific rows and columns by column names
display(df.loc[0:4, ["Plant_Name", "Plant_Part", "Solvent", "Extraction_Yield_pct"]])

# Select samples with extraction yield greater than 25%
display(df.loc[df["Extraction_Yield_pct"] > 25, ["Plant_Name", "Solvent", "Extraction_Yield_pct"]])

# Select samples with high antioxidant activity and display selected columns
display(df.loc[
    df["Antioxidant_Activity_pct"] >= 60,
    ["Plant_Name", "Total_Phenolic_Content_mg_GAE_g", "Antioxidant_Activity_pct", "Quality_Class"]
])
```

**Meaning of Code Block 7**

This code block introduces loc, a Pandas method used to select rows and columns by their labels or by conditions. Unlike iloc, which primarily uses integer positions, loc allows us to specify actual column names and apply logical conditions at the same time. Here, we select particular columns for selected rows, extract samples with an extraction yield above 25%, and identify samples with antioxidant activity of at least 60% while displaying only the relevant variables. loc is particularly useful for readable and precise data selection and will become important when we modify or preprocess specific parts of a dataset.

```python
# Block 8: Adding, Renaming, and Modifying Columns

# Create a new column with a calculated value
df["Phenolic_Flavonoid_Total"] = (
    df["Total_Phenolic_Content_mg_GAE_g"] +
    df["Total_Flavonoid_Content_mg_QE_g"]
)

# Create a new categorical column based on extraction yield
df["Yield_Category"] = pd.cut(
    df["Extraction_Yield_pct"],
    bins=[0, 15, 25, 50],
    labels=["Low", "Medium", "High"]
)

# Rename a column
df.rename(
    columns={"Temperature_C": "Extraction_Temperature_C"},
    inplace=True
)

# Display the modified DataFrame
display(df.head())
```

**Meaning of Code Block 8**

This code block demonstrates how to modify a Pandas DataFrame by creating new columns, generating categories from numerical values, and renaming existing columns. The Phenolic_Flavonoid_Total column is created by adding the total phenolic and flavonoid contents, while pd.cut() converts extraction yield into three categories—Low, Medium, and High—according to defined ranges. The rename() function changes the column name Temperature_C to Extraction_Temperature_C, and inplace=True applies the change directly to the existing DataFrame. Creating and modifying variables in this way is an important skill for preparing features and target variables for machine-learning models.

```python
# Block 9: Checking Missing Values and Duplicate Records

# Count missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# Calculate percentage of missing values in each column
print("\nPercentage of missing values:")
print((df.isnull().sum() / len(df) * 100).round(2))

# Count the total number of duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

# Display rows containing missing values
print("\nRows containing missing values:")
display(df[df.isnull().any(axis=1)])
```

**Meaning of Code Block 9**

This code block is used to identify common data-quality problems before machine-learning analysis. df.isnull().sum() counts the missing values in every column, while dividing this count by the total number of rows and multiplying by 100 calculates the percentage of missing values. df.duplicated().sum() identifies the number of completely duplicated rows, and df[df.isnull().any(axis=1)] displays all records containing at least one missing value. Detecting missing and duplicate data is an essential part of data preparation because unsuitable or incomplete records can affect statistical analysis and machine-learning model performance.

```python
# Block 10: Basic Statistical Summary of the Dataset

# Generate descriptive statistics for numerical columns
print("Statistical summary of numerical columns:")
display(df.describe())

# Generate descriptive statistics for all columns
print("\nStatistical summary of all columns:")
display(df.describe(include="all"))

# Calculate the mean of numerical columns
print("\nMean values:")
display(df.mean(numeric_only=True))

# Calculate the median of numerical columns
print("\nMedian values:")
display(df.median(numeric_only=True))
```

**Meaning of Code Block 10**

This code block performs basic statistical analysis of the dataset. df.describe() provides important descriptive statistics for numerical variables, including count, mean, standard deviation, minimum, quartiles, and maximum values, while df.describe(include="all") extends the summary to categorical columns as well. The mean() and median() functions calculate the average and middle values of numerical variables, respectively. These statistics help us understand the distribution, central tendency, variability, and general characteristics of the Pharmacognosy data before applying machine-learning algorithms.

```python
# Block 11: Counting and Summarizing Categorical Data

# Count unique values in selected categorical columns
print("Number of unique plant species:")
print(df["Plant_Name"].nunique())

print("\nNumber of unique extraction methods:")
print(df["Extraction_Method"].nunique())

# Display frequency of each plant species
print("\nPlant species frequency:")
display(df["Plant_Name"].value_counts())

# Display frequency of extraction methods
print("\nExtraction method frequency:")
display(df["Extraction_Method"].value_counts())

# Display frequency of quality classes
print("\nQuality class frequency:")
display(df["Quality_Class"].value_counts())
```

**Meaning of Code Block 11**

This code block is used to examine categorical variables and understand how frequently different categories occur in the dataset. The nunique() function counts the number of distinct values in a column, while value_counts() counts how many times each individual category appears. Here, we examine the number and distribution of plant species, extraction methods, and quality classes. Understanding categorical distributions is important because it helps identify the composition of the dataset and, later, determine whether classification targets are reasonably balanced or dominated by particular classes.

```python
# Block 12: Sorting and Ranking Data

# Sort the dataset by extraction yield in ascending order
df_sorted_ascending = df.sort_values(
    by="Extraction_Yield_pct",
    ascending=True
)

print("Samples with lowest extraction yield:")
display(df_sorted_ascending[[
    "Plant_Name",
    "Solvent",
    "Extraction_Method",
    "Extraction_Yield_pct"
]].head(10))

# Sort the dataset by extraction yield in descending order
df_sorted_descending = df.sort_values(
    by="Extraction_Yield_pct",
    ascending=False
)

print("\nSamples with highest extraction yield:")
display(df_sorted_descending[[
    "Plant_Name",
    "Solvent",
    "Extraction_Method",
    "Extraction_Yield_pct"
]].head(10))

# Sort by more than one column
print("\nSorted by extraction method and extraction yield:")
display(
    df.sort_values(
        by=["Extraction_Method", "Extraction_Yield_pct"],
        ascending=[True, False]
    )[[
        "Plant_Name",
        "Extraction_Method",
        "Extraction_Yield_pct"
    ]].head(15)
)
```

**Meaning of Code Block 12**

This code block demonstrates how to sort and rank records in a Pandas DataFrame. The sort_values() function arranges records according to one or more columns, while ascending=True sorts values from lowest to highest and ascending=False sorts them from highest to lowest. Here, we identify samples with the lowest and highest extraction yields and also demonstrate multi-column sorting by extraction method followed by extraction yield. Sorting is useful for quickly identifying high- or low-performing samples, comparing experimental conditions, and organizing data before further analysis.

```python
# Block 13: Grouping Data and Calculating Summary Statistics

# Calculate average extraction yield for each extraction method
print("Average extraction yield by extraction method:")
display(
    df.groupby("Extraction_Method")["Extraction_Yield_pct"]
      .mean()
      .sort_values(ascending=False)
)

# Calculate average antioxidant activity for each solvent
print("\nAverage antioxidant activity by solvent:")
display(
    df.groupby("Solvent")["Antioxidant_Activity_pct"]
      .mean()
      .sort_values(ascending=False)
)

# Calculate multiple statistics for extraction yield by plant part
print("\nExtraction yield statistics by plant part:")
display(
    df.groupby("Plant_Part")["Extraction_Yield_pct"]
      .agg(["count", "mean", "min", "max"])
      .sort_values("mean", ascending=False)
)
```

**Meaning of Code Block 13**

This code block demonstrates how to group records according to a categorical variable and calculate summary statistics for each group. The groupby() function divides the dataset into groups such as extraction methods, solvents, or plant parts, and functions such as mean() calculate the average value within each group. The agg() function allows several statistics, including count, mean, minimum, and maximum, to be calculated simultaneously. Grouping is particularly useful in Pharmacognosy because it allows us to compare outcomes such as extraction yield or antioxidant activity across different extraction methods, solvents, and plant materials.

```python
# Block 14: Removing and Restoring a Column

# Create a temporary copy of the DataFrame
df_temp = df.copy()

# Remove the temporary calculated column
df_temp.drop(columns=["Phenolic_Flavonoid_Total"], inplace=True)

print("Columns after removing Phenolic_Flavonoid_Total:")
print(df_temp.columns.tolist())

# Restore the calculated column
df_temp["Phenolic_Flavonoid_Total"] = (
    df_temp["Total_Phenolic_Content_mg_GAE_g"] +
    df_temp["Total_Flavonoid_Content_mg_QE_g"]
)

print("\nColumns after restoring the column:")
print(df_temp.columns.tolist())

display(df_temp.head())
```

**Meaning of Code Block 14**

This code block demonstrates how to remove and recreate a column in a Pandas DataFrame without permanently changing our original master dataset. df.copy() creates an independent copy called df_temp, and drop() removes the selected column from this temporary DataFrame. We then recreate the same column using the original numerical variables. Using a temporary copy is a good practice during learning because it allows us to experiment with DataFrame operations while protecting the original dataset that will be used throughout our machine-learning practicals.

```python
# Block 15: Working with DataFrame Copies

# Create a copy of the original DataFrame
df_copy = df.copy()

# Modify the copied DataFrame
df_copy["Extraction_Yield_pct"] = df_copy["Extraction_Yield_pct"].round(1)

# Compare the original and copied DataFrame
print("Original DataFrame:")
display(df[["Sample_ID", "Extraction_Yield_pct"]].head())

print("\nCopied DataFrame:")
display(df_copy[["Sample_ID", "Extraction_Yield_pct"]].head())

# Check whether both DataFrames are the same object
print("\nAre df and df_copy the same object?")
print(df is df_copy)
```

**Meaning of Code Block 15**

This code block demonstrates the importance of creating an independent copy of a DataFrame before making experimental changes. df.copy() creates a separate DataFrame called df_copy, allowing us to modify values without unintentionally changing the original df dataset. Here, the extraction-yield values in the copied DataFrame are rounded to one decimal place, while the original values remain unchanged. The expression df is df_copy checks whether both variables refer to the same Python object and returns False, confirming that the copy is independent. This practice is particularly important in machine-learning workflows where the original dataset should be preserved for reproducibility and comparison.

```python
# Block 16: Removing Duplicate Records and Resetting the Index

# Create a working copy of the DataFrame
df_clean = df.copy()

# Count duplicate records
duplicate_count = df_clean.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

# Remove duplicate records
df_clean = df_clean.drop_duplicates()

# Reset the DataFrame index after removing duplicates
df_clean = df_clean.reset_index(drop=True)

# Display the updated shape
print("Shape after removing duplicates:", df_clean.shape)

# Display the first five records
display(df_clean.head())
```

**Meaning of Code Block 16**

This code block demonstrates how to identify and remove duplicate records while preserving the original dataset. A copy of the DataFrame is first created as df_clean, after which duplicated().sum() counts completely duplicated rows. The drop_duplicates() function removes those duplicate records, and reset_index(drop=True) creates a new continuous index after the removal. Working with a separate copy is important because it protects the original master dataset and allows us to perform data-cleaning operations safely.

```python
# Block 17: Saving the DataFrame as an Excel File

# Save the current DataFrame as a new Excel file
output_file = "Pharmacognosy_ML_Processed_Data.xlsx"

df_clean.to_excel(output_file, index=False)

# Confirm that the file has been saved
print("File saved successfully as:", output_file)
```

**Meaning of Code Block 17**

This code block demonstrates how to export a Pandas DataFrame into an Excel file. The variable output_file stores the desired filename, and df_clean.to_excel() converts the DataFrame into an Excel workbook. The argument index=False prevents Pandas from adding the DataFrame index as an unnecessary additional column in the Excel file. Saving processed data is an important skill because it allows the cleaned or modified dataset to be reused for further analysis, documentation, and machine-learning workflows.

```python
# Block 18: Final Inspection of the DataFrame

print("Final dataset shape:")
print(df_clean.shape)

print("\nFinal column names:")
print(df_clean.columns.tolist())

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate records:")
print(df_clean.duplicated().sum())

print("\nFinal statistical summary:")
display(df_clean.describe())

print("\nFirst five records:")
display(df_clean.head())
```

**Meaning of Code Block 18**

This code block performs a final inspection of the working DataFrame before we proceed to subsequent machine-learning practicals. It checks the final number of rows and columns using shape, confirms the available variables using columns.tolist(), identifies remaining missing values using isnull().sum(), verifies whether duplicate records remain using duplicated().sum(), provides a statistical summary of numerical variables using describe(), and displays the first five records for a final visual check. This type of final dataset inspection is an important reproducibility and quality-control step before moving from basic Pandas operations to dedicated data preprocessing and machine-learning workflows.
