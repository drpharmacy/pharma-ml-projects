## Operators
Arithmetic \
Assignment \
Comparison \
Logical \
Identity \
Membership

```python
import math

a=20
b=6

print("Addition :",a+b)
print("Subtraction :",a-b)
print("Multiplication :",a*b)
print("Division :",a/b)   # 5/2 = 2.5
print("Floor Division :",a//b)
print("Ceil Division :",math.ceil(a/b))  #2.2, 2.5, 2.6 -> 3
print("Floor Division :",math.floor(a/b))
print("Modulus :",a%b)   #5%2 = 1
print("Power :",a**b)   #2**3 = 8


# 2.1, 2.2, .... 2.9 -> 2 & 3
# Ceil -> 3
# Floor -> 2
```

```python
x=10
print(x)

x += 5  # x = x + 5
print(x)

x -= 3   #x = x - 3
print(x)

x*=2
print(x)

x/=4
print(x)

x//=2
print(x)

x%=2
print(x)

x**=3
print(x)
```

```python
a=20
b=15

# Result -> Boolean (True/False)
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```
