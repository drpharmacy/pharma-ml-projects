```python
#dictionary create
Patient = { "ID":"P01",
            "Name":"Suchita",
            "Age":35
          }
Patient
```

```python
#dictionary Access Key only: value print
print(Patient["Name"])
print(Patient["ID"])
print(Patient["Age"])
```

```python
# Add key:value
Patient["weight"]=65
print(Patient)
```

```python

Patient["Blood Group"]='B'
print(Patient)
```

```python
#modify value
Patient['Name']="MOnika"
print(Patient)
```

```python
# delete key:value
del Patient['Blood Group']
print(Patient)
```

```python
#dispaly key value
Patient.values()
print(Patient)
```

```python
#clear dictionary {}
Patient.clear()
print(Patient)
```
