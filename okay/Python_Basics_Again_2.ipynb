```python
age = 20
if age >= 18:
    print("Eligible to Vote")
print("Done")
```

```python
temperature = 39
if temperature > 37:
    print("Patient has Fever")
```

```python
age = int(input("Age?:"))
if age >= 18:
    print("Eligible to Vote")
else:
    print("Not Eligible to vote")
```

```python
marks = int(input("Enter Marks : "))
if marks >= 90:
    print("Excellent")
elif marks >= 75:
    print("Good")
elif marks >= 50:
    print("Average")
else:
    print("Needs Improvement")
```

```python
bmi = float(input("Enter BMI : "))
if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Normal")
elif bmi < 30:
    print("Overweight")
else:
    print("Obese")
```

```python
age = int(input("Enter Age : "))
weight = float(input("Enter Weight : "))
if age >= 18:
    if weight >= 50:
        print("Eligible")
    else:
        print("Weight too low")
else:
    print("Age below 18")
```

```python
stock = int(input("Stock : "))
expiry = input("Expired? (Yes/No): ")

if stock > 0:
    if expiry == "No":
        print("Medicine Can Be Dispensed")
        print("Good to go.")
    else:
        print("Expired Medicine")
else:
    print("Out of Stock")
```
